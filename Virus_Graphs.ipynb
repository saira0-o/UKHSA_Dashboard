# Respiratory virus graphs

Use the buttons below to switch between the COVID-19, influenza, and RSV graphs.

```python
from pathlib import Path

import ipywidgets as widgets
from IPython.display import Image, clear_output, display

graphs = {
    "COVID-19": "covid_admissions_positivity.png",
    "Influenza": "influenza_admissions_positivity.png",
    "RSV": "RSV_cases_positivity.png",
}

virus_selector = widgets.ToggleButtons(
    options=list(graphs),
    description="Virus:",
    style={"description_width": "initial"},
)
graph_output = widgets.Output()

def show_graph(change):
    image_path = Path(graphs[change["new"]])
    if not image_path.is_file():
        raise FileNotFoundError(f"Graph image not found: {image_path}")
    with graph_output:
        clear_output(wait=True)
        display(Image(filename=str(image_path)))

virus_selector.observe(show_graph, names="value")
display(virus_selector, graph_output)
show_graph({"new": virus_selector.value})
```
