# Seasonal Positivity Charts

Choose a year to view seasonal positivity charts for RSV, influenza, and COVID-19. Selecting 2026 displays both observed data so far and the predicted forecast.

Season groupings: Spring (March–May), Summer (June–August), Autumn (September–November), and Winter (December–February).

```python
from pathlib import Path

import ipywidgets as widgets
from IPython.display import Image, clear_output, display

chart_files = {
    "2020": "2020_seasonal_positivity.png",
    "2021": "2021_seasonal_positivity.png",
    "2022": "2022_seasonal_positivity.png",
    "2023": "2023_seasonal_positivity.png",
    "2024": "2024_seasonal_positivity.png",
    "2025": "2025_seasonal_positivity.png",
}
forecast_file = "2026_seasonal_positivity_prediction.png"
observed_2026_file = "2026_seasonal_positivity.png"

year_selector = widgets.Dropdown(
    options=["2026", "2025", "2024", "2023", "2022", "2021", "2020"],
    value="2026",
    description="Year:",
    style={"description_width": "initial"},
)
chart_output = widgets.Output()


def make_chart_image(filename):
    image_path = Path(filename)
    if not image_path.is_file():
        raise FileNotFoundError(f"Chart image not found: {image_path}")
    return widgets.Image(
        value=image_path.read_bytes(),
        format="png",
        layout=widgets.Layout(width="100%", object_fit="contain"),
    )


def show_charts(change=None):
    with chart_output:
        clear_output(wait=True)
        if year_selector.value == "2026":
            display(widgets.HBox([
                widgets.VBox([
                    widgets.HTML("<h3>2026 so far</h3>"),
                    make_chart_image(observed_2026_file),
                ], layout=widgets.Layout(width="50%")),
                widgets.VBox([
                    widgets.HTML("<h3>2026 predicted forecast</h3>"),
                    make_chart_image(forecast_file),
                ], layout=widgets.Layout(width="50%")),
            ], layout=widgets.Layout(width="100%")))
        else:
            display(Image(filename=chart_files[year_selector.value]))


year_selector.observe(show_charts, names="value")
display(year_selector, chart_output)
show_charts()
```
